```python
df = pd.read_csv(path + "/online food delivery dataset.csv")
df.head()
```

```python
print("First 5 rows:")
print(df.head())

print("\nShape (rows, columns):", df.shape)

print("\nColumn names:", df.columns.tolist())

print("\nData types:\n", df.dtypes)
```

```python
avg_age = df['Age'].mean()
min_age = df['Age'].min()
max_age = df['Age'].max()
male_count = (df['Gender'] == 'Male').sum()
female_count = (df['Gender'] == 'Female').sum()

print("Average age:", avg_age)
print("Minimum age:", min_age)
print("Maximum age:", max_age)
print("Male customers:", male_count)
print("Female customers:", female_count)
```

```python
filtered = df[(df['Age'] > 25) & (df['Occupation'] == 'Employee')]
filtered[['Age', 'Gender', 'Occupation', 'Monthly Income']]
```

```python
avg_age_by_occupation = df.groupby('Occupation')['Age'].mean()
print(avg_age_by_occupation)

highest_occupation = avg_age_by_occupation.idxmax()
print("\nOccupation with highest average age:", highest_occupation)
```

```python
feedback_counts = df['Feedback'].value_counts()
print(feedback_counts)

positive_pct = (feedback_counts.get('Positive', 0) / len(df)) * 100
print("\nPercentage positive feedback:", round(positive_pct, 2), "%")
```

```python
class BankAccount:
    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        else:
            return False

acc = BankAccount(104, "Neha", 1000)

result1 = acc.withdraw(400)
print("Valid withdrawal result:", result1, "| Balance:", acc.balance)

result2 = acc.withdraw(2000)
print("Overdraw withdrawal result:", result2, "| Balance:", acc.balance)
```

```python
class Student:
    def __init__(self, name, rollno, marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks
    def average_marks(self):
        return sum(self.marks) / len(self.marks)


student1 = Student("siddhi", 16,[ 70,80,90])
student2 = Student("Alice", 18, [90, 85, 95])
student3 = Student("neha", 17, [88, 92, 85])

print(student1.name)
print(student2.name)
print(student3.name)


for s in [student1, student2, student3]:
    print(f"{s.name}'s average marks: {s.average_marks():.2f}")
```

```python
class Vehicle:
    def move(self):
        print("Vehicle is moving")

class Car(Vehicle):
    def move(self):
        print("Car is driving on the road")

class Boat(Vehicle):
    def move(self):
        print("Boat is sailing on water")

v = Vehicle()
c = Car()
b = Boat()

v.move()
c.move()
b.move()
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

car1 = Car("Honda", 4)
car2 = Car("Ferrari", 2)

print(f"{car1.brand} - {car1.doors} doors")
print(f"{car2.brand} - {car2.doors} doors")
```

```python
class Car:
    def move(self):
        print("Car drives on the road")

class Boat:
    def move(self):
        print("Boat sails on water")

class Airplane:
    def move(self):
        print("Airplane flies in the sky")

vehicles = [Car(), Boat(), Airplane()]

for vehicle in vehicles:
    vehicle.move()
```
