```python
marks = list(map(int, input("Enter marks of 5 subjects: ").split()))
average = sum(marks) / len(marks)

if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 40:
    grade = "D"
else:
    grade = "F"

print("Marks =", marks)
print("Average =", average)
print("Grade =", grade)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
vowels = "aeiouAEIOU"

num_words = len(words)
num_chars = len(sentence.replace(" ", ""))
num_vowels = 0
num_consonants = 0

for ch in sentence:
    if ch.isalpha():
        if ch in vowels:
            num_vowels += 1
        else:
            num_consonants += 1

print("Words =", num_words)
print("Charcters =", num_chars)
print("vowels =", num_vowels)
print("Consonant =", num_consonants)
```

```python
numbers = list(map(int, input("Enter numbers: ").split()))
print(numbers)

even = []
odd = []

for n in numbers:
    if n % 2 == 0:
        even.append(n)
    else:
        odd.append(n)

print("Even =", even)
print("odd =", odd)
```

```python
numbers = list(map(int, input("Enter numbers: ").split()))
print(numbers)

large = numbers[0]
small = numbers[0]
total = 0

for n in numbers:
    if n > large:
        large = n
    if n < small:
        small = n
    total = total + n

average = total / len(numbers)

print("Largest =", large)
print("Smallest =", small)
print("Sum =", total)
print("Average =", average)
```

```python
text = input("Enter a string: ")
print(text)

cleaned = text.replace(" ", "").lower()
reverse = cleaned[::-1]

if cleaned == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
marks = tuple(map(int, input("Enter student marks: ").split()))
print(marks)

highest = marks[0]
lowest = marks[0]
total = 0

for m in marks:
    if m > highest:
        highest = m
    if m < lowest:
        lowest = m
    total = total + m

class_average = total / len(marks)

above_average = 0
for m in marks:
    if m > class_average:
        above_average += 1

print("Highest =", highest)
print("Lowest =", lowest)
print("Class Average =", class_average)
print("Above Average Count =", above_average)
```

```python
n = int(input("How many students? "))
student_marks = {}

for i in range(n):
    name = input("Enter name: ")
    mark = int(input("Enter marks: "))
    student_marks[name] = mark

print(student_marks)

search = input("Enter name to search: ")

if search in student_marks:
    print(search, "marks =", student_marks[search])
else:
    print("Student not found")
```

```python
text = input("Enter a string: ")
freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

for ch, count in freq.items():
    print(ch, ":", count)
```

```python
set1 = set(input("Enter elements of set 1: ").split())
set2 = set(input("Enter elements of set 2: ").split())

print("Set1 =", set1)
print("Set2 =", set2)

print("Union =", set1 | set2)
print("Intersection =", set1 & set2)
print("Difference =", set1 - set2)
print("Symmetric Difference =", set1 ^ set2)
```

```python
contacts = {}

while True:
    print("\n1.Add 2.Search 3.Update 4.Delete 5.Display 6.Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter name: ")
        number = input("Enter number: ")
        contacts[name] = number
        print("Added")

    elif choice == "2":
        name = input("Enter name: ")
        if name in contacts:
            print(name, "=", contacts[name])
        else:
            print("Not found")

    elif choice == "3":
        name = input("Enter name: ")
        if name in contacts:
            contacts[name] = input("Enter new number: ")
            print("Updated")
        else:
            print("Not found")



        

    elif choice == "4":
        name = input("Enter name: ")
        if name in contacts:
            del contacts[name]
            print("Deleted")
        else:
            print("Not found")

    elif choice == "5":
        print(contacts)

    elif choice == "6":
        break

    else:
        print("Invalid choice")
```

```python
students = {}

while True:
    print("\n1.Add 2.Search 3.Update 4.Delete 5.Display 6.Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        roll = input("Enter roll number: ")
        name = input("Enter name: ")
        course = input("Enter course: ")
        marks = int(input("Enter marks: "))
        students[roll] = {"name": name, "course": course, "marks": marks}
        print("Added")

    elif choice == "2":
        roll = input("Enter roll number: ")
        if roll in students:
            print(students[roll])
        else:
            print("Not found")

    elif choice == "3":
        roll = input("Enter roll number: ")
        if roll in students:
            students[roll]["name"] = input("Enter new name: ")
            students[roll]["course"] = input("Enter new course: ")
            students[roll]["marks"] = int(input("Enter new marks: "))
            print("Updated")
        else:
            print("Not found")

    elif choice == "4":
        roll = input("Enter roll number: ")
        if roll in students:
            del students[roll]
            print("Deleted")
        else:
            print("Not found")

    elif choice == "5":
        for roll, info in students.items():
            print(roll, "=", info)

    elif choice == "6":
        break

    else:
        print("Invalid choice")
```
