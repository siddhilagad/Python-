```python
def analyze_temperatures(temps):
    temps = np.array(temps)
    avg = temps.mean()
    maximum = temps.max()
    minimum = temps.min()
    hottest_day = temps.argmax() + 1  # 1-indexed day
    above_avg = temps[temps > avg]
    return {
        'average': avg,
        'max': maximum,
        'min': minimum,
        'hottest_day': hottest_day,
        'above_average': above_avg
    }

temperatures = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])
result = analyze_temperatures(temperatures)
for k, v in result.items():
    print(f'{k}: {v}')
```

```python
import numpy as np

# Store only the numerical marks in a NumPy array
marks = np.array([
    [78, 82, 91],   # Student 1
    [65, 74, 80],   # Student 2
    [92, 88, 95],   # Student 3
    [56, 69, 72],   # Student 4
    [84, 90, 87]    # Student 5
])

subjects = ['Mathematics', 'Physics', 'Programming']

# Average marks of each student (row-wise mean)
avg_per_student = marks.mean(axis=1)

# Average marks in each subject (column-wise mean)
avg_per_subject = marks.mean(axis=0)

# Highest and lowest mark in the entire dataset
highest_mark = marks.max()
lowest_mark = marks.min()

# Student with the highest average
best_student_index = avg_per_student.argmax()

# Subject with the highest average
best_subject_index = avg_per_subject.argmax()

# Display results
print("Marks matrix:\n", marks)
print("\nAverage marks of each student:", avg_per_student)
print("Average marks in each subject:", avg_per_subject)
print("\nHighest mark in the dataset:", highest_mark)
print("Lowest mark in the dataset:", lowest_mark)
print(f"\nStudent with the highest average: Student {best_student_index + 1} "
      f"(avg = {avg_per_student[best_student_index]:.2f})")
print(f"Subject with the highest average: {subjects[best_subject_index]} "
      f"(avg = {avg_per_subject[best_subject_index]:.2f})")
```

```python
def transform_measurements(data):
    data = np.array(data)
    squared = data ** 2
    sqrt_vals = np.sqrt(data)
    exp_vals = np.exp(data)
    mean_transformed = {
        'mean_of_squares': squared.mean(),
        'mean_of_sqrt': sqrt_vals.mean(),
        'mean_of_exp': exp_vals.mean()
    }
    return squared, sqrt_vals, exp_vals, mean_transformed

measurements = [2, 4, 6, 8, 10, 12]
squared, sqrt_vals, exp_vals, means = transform_measurements(measurements)

print('Squares:', squared)
print('Square roots:', sqrt_vals)
print('Exponentials:', exp_vals)
print('Means of transformed data:', means)
```

```python
import numpy as np

# Measurements from the two experiments
experiment_A = np.array([12, 15, 14, 18, 20, 17, 16])
experiment_B = np.array([10, 13, 15, 16, 19, 21, 18])

# Mean of each experiment
mean_A = experiment_A.mean()
mean_B = experiment_B.mean()

# Standard deviation of each experiment
std_A = experiment_A.std()
std_B = experiment_B.std()

# Maximum value of each experiment
max_A = experiment_A.max()
max_B = experiment_B.max()

# Minimum value of each experiment
min_A = experiment_A.min()
min_B = experiment_B.min()

# Difference between the two experiments for each measurement
difference = experiment_A - experiment_B

# Display results
print(f"Mean       -> A: {mean_A:.2f}, B: {mean_B:.2f}")
print(f"Std Dev    -> A: {std_A:.2f}, B: {std_B:.2f}")
print(f"Max        -> A: {max_A}, B: {max_B}")
print(f"Min        -> A: {min_A}, B: {min_B}")
print(f"Difference (A - B): {difference}")

# Determine which experiment shows greater variation
if std_A > std_B:
    print(f"\nExperiment")
```

```python
def generate_dataset(rows=10, cols=5, low=0, high=100):
    return np.random.randint(low, high, size=(rows, cols))

def analyze_dataset(dataset):
    mean_per_variable = dataset.mean(axis=0)
    mean_per_experiment = dataset.mean(axis=1)
    overall_max = dataset.max()
    overall_min = dataset.min()
    overall_mean = dataset.mean()
    above_mean = dataset[dataset > overall_mean]
    return {
        'mean_per_variable': mean_per_variable,
        'mean_per_experiment': mean_per_experiment,
        'max': overall_max,
        'min': overall_min,
        'above_mean': above_mean
    }

dataset = generate_dataset()
print('Dataset:\n', dataset)
print('Shape:', dataset.shape, '| Dimensions:', dataset.ndim)

analysis = analyze_dataset(dataset)
for key, value in analysis.items():
    print(f'{key}: {value}')

reshaped = dataset.reshape(5, 10)
print('Reshaped (5x10):\n', reshaped)

flattened = reshaped.flatten()
print('Flattened back to 1D:', flattened)
```

```python
def generate_marks(num_students=10, num_subjects=4):
    return np.random.randint(0, 101, size=(num_students, num_subjects))

def assign_grade(average):
    if average >= 90:
        return 'A'
    elif average >= 75:
        return 'B'
    elif average >= 60:
        return 'C'
    elif average >= 40:
        return 'D'
    else:
        return 'F'

def analyze_performance(marks):
    student_avg = marks.mean(axis=1)
    subject_avg = marks.mean(axis=0)
    best_student = student_avg.argmax()
    worst_student = student_avg.argmin()
    highest_mark = marks.max()
    lowest_mark = marks.min()
    count_above_75 = (student_avg > 75).sum()
    marks_above_90 = marks[marks > 90]
    return {
        'student_avg': student_avg,
        'subject_avg': subject_avg,
        'best_student': best_student,
        'worst_student': worst_student,
        'highest_mark': highest_mark,
        'lowest_mark': lowest_mark,
        'count_above_75': count_above_75,
        'marks_above_90': marks_above_90
    }

def build_student_records(marks, subjects):
    records = []
    for i, row in enumerate(marks):
        avg = row.mean()
        record = {
            'student': f'Student {i + 1}',
            'marks': dict(zip(subjects, row.tolist())),
            'average': round(float(avg), 2),
            'grade': assign_grade(avg)
        }
        records.append(record)
    return records

subjects = ['Math', 'Physics', 'Chemistry', 'Programming']
marks = generate_marks()
results = analyze_performance(marks)
records = build_student_records(marks, subjects)

print('Marks matrix:\n', marks)
print()
print('--- Analysis ---')
print('Average per student:', results['student_avg'])
print('Average per subject:', dict(zip(subjects, results['subject_avg'])))
print(f"Best student: Student {results['best_student'] + 1}")
print(f"Worst student: Student {results['worst_student'] + 1}")
print('Highest mark overall:', results['highest_mark'])
print('Lowest mark overall:', results['lowest_mark'])
print('Students with average > 75:', results['count_above_75'])
print('All marks > 90:', results['marks_above_90'])
print()
print('--- Student Report Card ---')
for record in records:
    print(f"{record['student']}: avg={record['average']}, grade={record['grade']}, marks={record['marks']}")
```
