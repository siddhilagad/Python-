```python
#theroy Question answer in Latex

\documentclass[11pt,a4paper]{article}
\usepackage[margin=1in]{geometry}
\usepackage{hyperref}
\usepackage{xcolor}
\usepackage{listings}
\usepackage{enumitem}
\usepackage{titlesec}
\usepackage{parskip}

\hypersetup{
    colorlinks=true,
    linkcolor=blue,
    urlcolor=blue
}

\lstdefinestyle{pystyle}{
    language=Python,
    basicstyle=\ttfamily\small,
    keywordstyle=\color{blue},
    stringstyle=\color{orange!70!black},
    commentstyle=\color{gray},
    numbers=left,
    numberstyle=\tiny\color{gray},
    frame=single,
    breaklines=true,
    showstringspaces=false,
    backgroundcolor=\color{gray!5}
}
\lstset{style=pystyle}

\titleformat{\section}{\large\bfseries}{\thesection}{1em}{}
\titleformat{\subsection}{\normalsize\bfseries}{\thesubsection}{1em}{}

\title{\textbf{Python Programming \\ Assignment 1}}
\author{Name: \underline{\hspace{4cm}} \\ Roll Number: \underline{\hspace{4cm}}}
\date{2nd Aug 2026}

\begin{document}
\maketitle

\noindent \textbf{GitHub Repository:} \href{https://github.com/username/repository}{GitHub Repository}
\textit{(replace the link above with your own repository URL before submitting)}

\vspace{0.5cm}
\noindent\textbf{Note:} Every mini-experiment's code must actually be run in Jupyter Notebook by you, and a screenshot of the output pasted below the corresponding explanation, as required by the submission guidelines. This report gives you the code, the expected output, and the explanation in your own words as a starting point.

\tableofcontents
\newpage

\section{Variables}
\subsection*{What is it?}
A variable is a named label that refers to a value stored in memory. In Python, a variable is created the moment a value is assigned to a name, using the \texttt{=} operator, e.g. \texttt{x = 10}. Python variables do not need an explicit type declaration; the type is decided automatically based on the value assigned.

\subsection*{Why do we need it?}
\begin{itemize}[nosep]
    \item Variables let us store data temporarily so it can be reused, updated, and referenced by a readable name instead of a raw memory address.
    \item They make code readable and maintainable -- a name like \texttt{age} is far clearer than an unnamed literal value repeated everywhere.
    \item They allow programs to work with changing data (user input, calculations, loop counters) rather than fixed, hard-coded values.
\end{itemize}

\subsection*{Where and How is it used?}
\begin{itemize}[nosep]
    \item Used everywhere in a program: to hold user input, intermediate results of calculations, loop counters, configuration values, etc.
    \item Created simply by assignment: \lstinline{name = "Alice"}, \lstinline{marks = 91.5}.
    \item Python allows multiple assignment in one line: \lstinline{a, b, c = 1, 2, 3}.
    \item Variable names follow rules: they must start with a letter or underscore, can contain letters, digits, and underscores, and are case-sensitive.
\end{itemize}

\section{Basic Data Types}

\subsection{Integer}
\textbf{What:} A whole number, positive or negative, without a decimal point (e.g. \texttt{5}, \texttt{-120}). In Python the type is \texttt{int}, and it has no fixed size limit (unlike languages such as C).

\textbf{Why:} Needed to represent countable, discrete quantities -- counters, indices, ages, quantities -- where fractional values do not make sense.

\textbf{Where/How:} Created directly by writing a whole number: \lstinline{count = 10}. Used in loops, indexing, arithmetic, and any situation involving whole-number counts.

\subsection{Float}
\textbf{What:} A number with a decimal point, used to represent real numbers (e.g. \texttt{3.14}, \texttt{-0.5}). The type is \texttt{float}.

\textbf{Why:} Needed whenever precision beyond whole numbers is required -- measurements, prices, averages, scientific calculations.

\textbf{Where/How:} Created by writing a number with a decimal point: \lstinline{price = 99.99}, or by dividing two integers with \texttt{/}, which always returns a float in Python 3.

\subsection{Boolean}
\textbf{What:} A data type with only two possible values: \texttt{True} or \texttt{False}. The type is \texttt{bool}, and internally it is a subtype of \texttt{int} (\texttt{True} = 1, \texttt{False} = 0).

\textbf{Why:} Needed to represent logical states and to control decision-making in a program -- whether a condition holds or not.

\textbf{Where/How:} Produced automatically by comparison operators (\lstinline{5 > 3} gives \texttt{True}), and used directly in \texttt{if} statements, \texttt{while} loops, and logical operators (\texttt{and}, \texttt{or}, \texttt{not}).

\subsection{String}
\textbf{What:} A sequence of characters enclosed in single, double, or triple quotes (e.g. \texttt{"Hello"}). The type is \texttt{str}, and strings are immutable.

\textbf{Why:} Needed to represent and manipulate text -- names, messages, file contents, user input.

\textbf{Where/How:} Created with quotes: \lstinline{name = "Python"}. Supports indexing, slicing, concatenation (\texttt{+}), repetition (\texttt{*}), and many built-in methods (\texttt{.upper()}, \texttt{.split()}, etc.).

\section{Arithmetic Operators}
\subsection*{What is it?}
Symbols used to perform mathematical calculations on numeric values: \texttt{+} (addition), \texttt{-} (subtraction), \texttt{*} (multiplication), \texttt{/} (true division), \texttt{//} (floor division), \texttt{\%} (modulus/remainder), and \texttt{**} (exponentiation).

\subsection*{Why do we need it?}
Almost every program involves some numeric computation -- totals, averages, discounts, positions, physics calculations -- and arithmetic operators are the basic building blocks for that.

\subsection*{Where and How is it used?}
\begin{itemize}[nosep]
    \item Directly between two numeric operands: \lstinline{total = price * quantity}.
    \item \texttt{//} is used when only the whole-number part of a division is needed (e.g. splitting items into equal groups).
    \item \texttt{\%} is commonly used to check divisibility (e.g. even/odd checks: \lstinline{n % 2 == 0}) or to wrap values within a range.
    \item \texttt{**} is used for powers and roots, e.g. \lstinline{2 ** 0.5} for square root.
\end{itemize}

\section{Comparison Operators}
\subsection*{What is it?}
Operators that compare two values and return a Boolean result: \texttt{==} (equal), \texttt{!=} (not equal), \texttt{>}, \texttt{<}, \texttt{>=}, \texttt{<=}.

\subsection*{Why do we need it?}
They let a program make decisions by testing relationships between values -- essential for conditions, filtering data, and controlling loops.

\subsection*{Where and How is it used?}
\begin{itemize}[nosep]
    \item Used inside \texttt{if}/\texttt{elif}/\texttt{while} conditions: \lstinline{if age >= 18:}.
    \item Used to compare numbers, strings (lexicographically), or other comparable objects.
    \item Can be chained in Python: \lstinline{0 < x < 100} checks both conditions at once.
\end{itemize}

\section{Assignment Operators}
\subsection*{What is it?}
Operators that assign a value to a variable, including the plain \texttt{=} and the compound forms \texttt{+=}, \texttt{-=}, \texttt{*=}, \texttt{/=}, \texttt{//=}, \texttt{\%=}, \texttt{**=}, which update a variable based on its current value.

\subsection*{Why do we need it?}
They provide a short, readable way to update a variable's value, especially useful in loops and running totals, instead of writing \lstinline{x = x + 1} repeatedly.

\subsection*{Where and How is it used?}
\begin{itemize}[nosep]
    \item \lstinline{score = 0} followed by \lstinline{score += 10} inside a loop to accumulate a total.
    \item Any compound operator combines the arithmetic operation with assignment: \lstinline{x *= 2} doubles \texttt{x} and stores the result back in \texttt{x}.
\end{itemize}

\section{Type Conversion}
\subsection*{What is it?}
The process of converting a value from one data type to another, e.g. string to integer using \texttt{int()}, integer to string using \texttt{str()}, string to float using \texttt{float()}. It can be implicit (done automatically by Python) or explicit (done manually by the programmer, also called type casting).

\subsection*{Why do we need it?}
Data does not always arrive in the type we need -- for example, \texttt{input()} always returns a string, so numeric input must be explicitly converted before arithmetic can be performed on it. Conversion prevents type errors and enables correct operations.

\subsection*{Where and How is it used?}
\begin{itemize}[nosep]
    \item \lstinline{age = int(input("Enter age: "))} converts user input text into an integer.
    \item \lstinline{str(25)} converts a number into text so it can be concatenated with other strings.
    \item Implicit conversion happens automatically, e.g. \lstinline{3 + 4.0} implicitly converts \texttt{3} to a float, giving \texttt{7.0}.
\end{itemize}

\section{Input and Output}
\subsection{\texttt{input()}}
\textbf{What:} A built-in function that pauses program execution and reads a line of text typed by the user from the keyboard. It always returns the value as a \texttt{str}.

\textbf{Why:} Needed to make programs interactive -- to receive data from the user at run time rather than hard-coding values.

\textbf{Where/How:} \lstinline{name = input("Enter your name: ")}. The optional argument is a prompt message shown to the user. If a number is needed, the result must be explicitly converted, e.g. \lstinline{int(input(...))}.

\subsection{\texttt{print()}}
\textbf{What:} A built-in function that displays output to the console/screen.

\textbf{Why:} Needed to show results, messages, and program status to the user.

\textbf{Where/How:} \lstinline{print("Hello", name)}. Supports multiple comma-separated arguments (joined by a space by default), and keyword arguments like \texttt{sep} and \texttt{end} to control formatting.

\section{Strings}
\subsection{String Creation}
\textbf{What:} Making a string value using single quotes \texttt{'...'}, double quotes \texttt{"..."}, or triple quotes \texttt{'''...'''}/\texttt{"""..."""} for multi-line text.

\textbf{Why:} Needed to represent any textual data in a program.

\textbf{Where/How:} \lstinline{s = "Hello World"}. Triple quotes are used for strings that span multiple lines or for docstrings.

\subsection{Indexing}
\textbf{What:} Accessing a single character of a string by its position. Indexing starts at \texttt{0} from the left, and negative indices count from the end (\texttt{-1} is the last character).

\textbf{Why:} Needed to read or extract a specific character from text.

\textbf{Where/How:} \lstinline{s[0]} gives the first character, \lstinline{s[-1]} gives the last character.

\subsection{Slicing}
\textbf{What:} Extracting a sub-part (substring) of a string using the syntax \lstinline{s[start:stop:step]}, where \texttt{start} is inclusive and \texttt{stop} is exclusive.

\textbf{Why:} Needed to extract portions of text without manually looping through characters.

\textbf{Where/How:} \lstinline{s[0:5]} gives the first 5 characters; \lstinline{s[::-1]} reverses the string by using a step of \texttt{-1}.

\subsection{String Operations}
\textbf{What:} Operations such as concatenation (\texttt{+}), repetition (\texttt{*}), membership testing (\texttt{in}), length (\texttt{len()}), and built-in methods (\texttt{.upper()}, \texttt{.lower()}, \texttt{.strip()}, \texttt{.split()}, \texttt{.replace()}, etc.).

\textbf{Why:} Needed for common text-processing tasks -- joining names, formatting output, cleaning user input, searching text.

\textbf{Where/How:} \lstinline{"Hello" + " " + "World"} concatenates; \lstinline{"Hi" * 3} repeats; \lstinline{"a" in "cat"} checks membership; \lstinline{len("Hello")} returns \texttt{5}.

\section{Lists}
\subsection{Creating Lists}
\textbf{What:} A list is an ordered, mutable collection of items, which can be of different types, created using square brackets \texttt{[...]}.

\textbf{Why:} Needed to store and manage a collection of related values together, rather than creating a separate variable for each one.

\textbf{Where/How:} \lstinline{fruits = ["Apple", "Banana", "Mango"]}.

\subsection{Indexing}
\textbf{What:} Accessing a single element of a list by its position, using zero-based indices (or negative indices from the end), the same way as string indexing.

\textbf{Why:} Needed to read or modify a specific element of the collection.

\textbf{Where/How:} \lstinline{fruits[0]} gives the first element, \lstinline{fruits[-1]} gives the last.

\subsection{Slicing}
\textbf{What:} Extracting a sub-list using \lstinline{list[start:stop:step]}, following the same rules as string slicing.

\textbf{Why:} Needed to work with a portion of a list without altering the original.

\textbf{Where/How:} \lstinline{fruits[1:3]} gives elements at index 1 and 2 (a new list).

\subsection{Updating Elements}
\textbf{What:} Changing the value stored at a particular index of a list, since lists are mutable.

\textbf{Why:} Needed to modify data in place -- for example, correcting a value or updating a score -- without creating a whole new list.

\textbf{Where/How:} \lstinline{numbers[1] = 200} replaces the element at index 1.

\subsection{Nested Lists}
\textbf{What:} A list that contains other lists as its elements, used to represent multi-dimensional data such as a grid, matrix, or table.

\textbf{Why:} Needed to model hierarchical or tabular data, e.g. a 2D matrix or a list of student records.

\textbf{Where/How:} \lstinline{matrix = [[1, 2], [3, 4]]}; an inner element is accessed with two indices, e.g. \lstinline{matrix[0][1]} gives \texttt{2}.

\subsection{Basic List Methods}
\textbf{What:} Built-in functions that operate on lists to add, remove, or reorganize elements.

\textbf{Why:} Needed to manipulate list contents conveniently without writing manual loops.

\textbf{Where/How:}
\begin{itemize}[nosep]
    \item \lstinline{append(x)} -- adds \texttt{x} to the end of the list.
    \item \lstinline{insert(i, x)} -- inserts \texttt{x} at index \texttt{i}, shifting later elements right.
    \item \lstinline{remove(x)} -- removes the first occurrence of value \texttt{x}.
    \item \lstinline{pop(i)} -- removes and returns the element at index \texttt{i} (last element if no index given).
    \item \lstinline{sort()} -- sorts the list in place in ascending order (or by a given key).
    \item \lstinline{reverse()} -- reverses the order of elements in place.
    \item \lstinline{len(list)} -- returns the number of elements in the list (a built-in function, not a method).
\end{itemize}





```
